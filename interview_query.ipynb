```python
#n grams
from collections import defaultdict

def get_ngrams(n,string):
    ngrams = defaultdict(int)
    out_of_bounds = len(string) - 1
    for i in range(len(string)):
        k = n
        our_string = string[i]
        j = 1
        while (k > 0):
            index = i + j
            if index <= out_of_bounds:
                our_string += string[i+j]
                j += 1
                k -= 1
            else:
                return dict(ngrams)
        #print(our_string)
        ngrams[our_string] += 1
        print(dict(ngrams))

    return dict(ngrams)

get_ngrams(2, 'banana')
```

```python
# over 100 dollars

import pandas as pd

transactions = {"transaction_id" : [1, 2, 3, 4, 5], "product_id" : [101, 102, 103, 104, 105], "amount" : [3, 5, 8, 3, 2]}

products = {"product_id" : [101, 102, 103, 104, 105], "price" : [20.00, 21.00, 15.00, 16.00, 52.00]}

df_transactions = pd.DataFrame(transactions)

df_products = pd.DataFrame(products)

def transactions_over_100(df_transactions: pd.DataFrame, df_products: pd.DataFrame):
    i = len(df_transactions['transaction_id'])
    j = 0
    df_transactions_over_100 = pd.DataFrame(columns=['transaction_id','product_id','amount','total_value'])
    while j < i:
      id = df_transactions['transaction_id'][j]
      product_id = df_transactions['product_id'][j]
      amount = df_transactions['amount'][j]
      price_index = df_products['product_id'].to_list().index(product_id)
      price = df_products['price'][price_index]
      total = amount * price
      if total > 100:
          series = pd.Series({'transaction_id': id, 'product_id': product_id, 'amount': amount, 'total_value': total})
          df_transactions_over_100.loc[len(df_transactions_over_100)] = series
      j += 1
    print(df_transactions_over_100)
    return df_transactions_over_100

    transactions_over_100(df_transactions, df_products)

```

```python
# first non-repeating character

string = 'happymeal'

def first_uniq_char(s):
    set_of_chars = []
    set_of_chars_that_repeat = []
    for a in s:
        if a in set_of_chars:
            set_of_chars_that_repeat.append(a)
        else:
            set_of_chars.append(a)
            pass
    list_of_nonrepeating_chars =  [a for a in set_of_chars if a not in set_of_chars_that_repeat]
    print(list_of_nonrepeating_chars)
    if list_of_nonrepeating_chars is None or len(list_of_nonrepeating_chars) < 1:
        return -1
    else:
        return s.index(list_of_nonrepeating_chars[0])

a = first_uniq_char(string)
print(a)
```

```python
#std of list

import pandas as pd
from collections import defaultdict

def compute_deviation(list_numbers):
    std = defaultdict(float)
    for a in list_numbers:
        s = pd.Series(a['values'])
        key = a['key']
        # s = pd.Series(values)
        intt = s.std()
        std[key] = intt

    print(std)


key_dict = [
    {
        'key': 'list1',
        'values': [4,5,2,3,4,5,2,3],
    },
    {
        'key': 'list2',
        'values': [1,1,34,12,40,3,9,7],
    }
]

compute_deviation(key_dict)
```
